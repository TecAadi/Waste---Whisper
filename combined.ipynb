```python
import os
import shutil

# Main dataset location
BASE_DIR = r"C:\Users\AADI  SINGH\OneDrive\Desktop\4 year Project\kaggle waste(1)"

# Two folders you want to combine
FOLDER1 = os.path.join(BASE_DIR, "original")
FOLDER2 = os.path.join(BASE_DIR, "standardized_256")

# New combined folder
COMBINED_DIR = os.path.join(BASE_DIR, "combined")

# Create combined folder
os.makedirs(COMBINED_DIR, exist_ok=True)

# Same class folders in both datasets
classes = [
    "battery",
    "biological",
    "cardboard",
    "clothes",
    "glass",
    "metal",
    "paper",
    "plastic",
    "shoes",
    "trash"
]

# Combine each class
for class_name in classes:

    # Destination class folder
    destination = os.path.join(COMBINED_DIR, class_name)
    os.makedirs(destination, exist_ok=True)

    # -------- ORIGINAL --------
    source1 = os.path.join(FOLDER1, class_name)

    if os.path.exists(source1):
        for file in os.listdir(source1):

            source_file = os.path.join(source1, file)

            if os.path.isfile(source_file):
                destination_file = os.path.join(
                    destination,
                    "original_" + file
                )

                shutil.copy2(source_file, destination_file)

    # -------- STANDARDIZED 256 --------
    source2 = os.path.join(FOLDER2, class_name)

    if os.path.exists(source2):
        for file in os.listdir(source2):

            source_file = os.path.join(source2, file)

            if os.path.isfile(source_file):
                destination_file = os.path.join(
                    destination,
                    "standardized256_" + file
                )

                shutil.copy2(source_file, destination_file)

print("✅ Both datasets combined successfully!")
```
